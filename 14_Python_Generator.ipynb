# **Generator is a simple way of creating iterator. instead of return to everything at a time, generator use yield to give one value at a time. Generators are usefull for handling large data because it is a memory efficient function.**

```python
def cupcakes_yield():
    yield 1
    yield 2
    yield 4
    yield 5

for cake in cupcakes_yield():
    print(cake)
```

```python
#return
def cupcakes_return():
    return [1, 2, 3, 4, 5]  # all at once

print(cupcakes_return())  # [1, 2, 3, 4, 5]
```
