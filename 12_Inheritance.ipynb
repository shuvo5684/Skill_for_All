#**Inheritance means a class can get attributes and methods from another class.the class that gives features is called parent/super/base class and the class that get features is called child/sub/ derrived class.**#

```python
#paren class
class Animal:
  def __init__(self,name):
    self.name=name
  def speak(self):
    print(f"{self.name} makes a sound")
#child class
class Dog(Animal):
  def speak(self):
    print(f"{self.name} barks")


# lets make a objects now
animal=Animal("generic animal")
dog=Dog("Buddy")
animal.speak()
dog.speak()
```
