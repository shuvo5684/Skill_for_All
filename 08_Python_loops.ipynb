# **In Python, there are two loops, such as i) for loop and ii) While loop**

## **1. for loop**

```python
fruits=["Apple","Banana","Cherry"]
for x in fruits:
    print(x)
```

```python
for i in range(5):
  if i==3:
    continue
  if i==4:
    break
  print(i)
```

```python
for i in range(2,20,2):
  print(i)
```

## **2. While loop**

```python
count=1
while count<=5:
  print("shuvo")
  count +=1
```

```python
while True:
  name=input("Enter your name (Enter stop end the loop)")
  if name=="stop":
    break
  print("Hello", name)
```
