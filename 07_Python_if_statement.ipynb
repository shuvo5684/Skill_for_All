```python
num=input("Enter the number: ")
num=int(num)

if num%2==0:
  print("the number is even and the number is: ",num)
else:
  print("the number is odd and the number is: ",num)
```

```python
Bangladeshi=["vat","dal","mas","Mangso"]
indian=["porata","ruti","samosa"]
italian=["pizza","pasta"]

dish=input("Enter the dish name: ")
if dish in Bangladeshi:
    print("Bangladeshi dish")
elif dish in indian:
    print("Indian dish")
elif dish in italian:
    print("Italian dish")
else:
    print("Not a dish")
```
