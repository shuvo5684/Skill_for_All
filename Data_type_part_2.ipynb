# **List is ordered and changeable**

```python
fruits=['apple','banana','cherry']
print(fruits)
```

```python
fruits[1]='kiwi' # banana replaced by 'kiwi'
```

```python
print(fruits)
```

```python
fruits.append('orange') # orange added to the list as a last item.
```

```python
print('updated list:',fruits)
```
