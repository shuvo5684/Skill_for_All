**Write a program to open a bank account, deposit and withdraw amount and send message to the account holder to current status of an account.Using class and objects of a python**

```python
class BankAccount: # defining a class as a BankAccount
  Bank_Name="City_Bank"
  def __init__(self,owner,balance):
    self.owner=owner
    self.balance=balance
    print("Account Holder Name:",self.owner)
    print("Account Balance:",self.balance)

  def deposit(self,amount): #Method 1 to deposit amount
    self.balance +=amount
    print(f"Name: {self.owner}:\n deposited amount {amount} \n total banance is {self.balance}")

  def withdraw(self,amount): #method 2 withdraw amount
    self.balance -=amount
    print(f"Name: {self.owner}:\n withdrawn amount {amount} \n total banance is {self.balance}")


# Creating a objects from the Class
Account_holder_1= BankAccount("Rahim",10000)
Account_holder_2=BankAccount("Karim",20000)

#Now objects using class methods

Account_holder_1.deposit(5000)
Account_holder_2.withdraw(3000)
```
