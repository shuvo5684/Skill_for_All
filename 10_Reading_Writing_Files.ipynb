```python
#Writing file
with open("test.txt","w") as file:
  file.write("I Love Python\n")
  file.write("this files for python writings\n")
```

```python
#Read entire file
with open("test.txt","r") as file:
  content=file.read()
  print(content)
```

```python
with open("test.txt", "r") as file:
  print(file.read())
```

```python
#read file line by line
with open("test.txt","r") as file:
  for line in file:
    print(line.strip())
```

```python
#reading file as a list
with open("test.txt","r") as file:
  lines=file.readlines()
  print(lines)
```

```python
#appending a file
with open("test.txt","a") as file:
  file.write("This is a new line\n")
```

```python
#reading again
with open("test.txt","r") as file:
  print(file.read())
```

```python
#writing  files
with open("test.bin","wb") as file:
  file.write(b"This is a binary file")
```

```python
#reading binary
with open("test.bin","rb") as file:
  print(file.read())
```
