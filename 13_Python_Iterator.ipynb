#**An Iterstor is an object used to traverse through all the elements of a collection(list,tuples,dictionaries) one element at a time. It follows the iterator protocols which involves two key methods one is __iter__(): returns the iterator object itself and other one is __next__(): returns the next value from the sequence and raise stopiteration when the sequence ends.**

```python
numbers=[1,2,3,4,5]
it=iter(numbers)
print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
```

```python
for i in numbers:
  print(i)
```

# **Creating a custom iterator. some steps to be followed. 1. Define a class that act as a iterator. 2. initialise attribute. 3. implement iter: returns itself 4. implement next:should provide the next item in the sequence each timme its call.**

```python
class count:
  def __init__(self,start,end=50):
    self.start=start
    self.end=end

  def __iter__(self):
    return self
  def __next__(self):
    if self.start<=self.end:
      current=self.start
      self.start +=1
      return current
    else:
      raise StopIteration

#using the custom iterator
for i in count(40):
    print(i)
```
